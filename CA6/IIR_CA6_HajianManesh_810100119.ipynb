<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">

🎯 <b>خروجی مورد انتظار:</b><br>
دیتافریم از اسناد پارس شده به همراه متادیتای آن‌ها

```python
PAPERS_DIR = Path("papers")
pdf_paths = sorted(PAPERS_DIR.rglob("*.pdf"))
assert pdf_paths, f"No PDFs found in: {PAPERS_DIR.resolve()}"
print("✅ PDFs found:", len(pdf_paths))



def read_first_page_markdown(pdf_path: Path) -> str:
    """Convert only page 0 of a PDF into markdown text."""
    md = pymupdf4llm.to_markdown(str(pdf_path), pages=[0])
    return (md or "").strip()

def normalize_markdown_text(s: str) -> str:
    """Remove common markdown styling and normalize whitespace."""
    s = (s or "").strip()
    s = re.sub(r"\*\*(.*?)\*\*", r"\1", s)          # **bold** -> bold
    s = re.sub(r"_([^_]+)_", r"\1", s)              # _italics_ -> italics
    s = re.sub(r"\s+", " ", s)                      # collapse whitespace
    return s.strip()

BAD_LINE_PREFIXES = (
    "abstract", "keywords", "introduction", "contents",
    "proceedings", "copyright", "permission", "provided proper attribution",
    "this work was published", "marginally revised version"
)
BAD_LINE_CONTAINS = ("@", "http", "https", "doi:", "arxiv")

def is_title_like(line: str) -> bool:
    if not line:
        return False

    raw = line.strip()
    low = raw.lower()

    if len(raw) < 8 or len(raw) > 180:
        return False

    if any(low.startswith(p) for p in BAD_LINE_PREFIXES):
        return False
    if any(x in low for x in BAD_LINE_CONTAINS):
        return False

    # require some letters
    if sum(ch.isalpha() for ch in raw) < 6:
        return False

    # reject pure numeric/symbol lines
    if re.fullmatch(r"[\W\d_]+", raw):
        return False

    return True

def extract_title_from_md(md_text: str) -> str | None:
    """
    Same priority logic as your friend's code:
    1) Prefer '## ...'
    2) Then accept '# ...'
    3) Then bold-only line (**...**)   (kept same intention)
    4) Else: if first line is plausible, accept it
    """
    if not md_text:
        return None

    lines = [ln.strip() for ln in md_text.splitlines() if ln.strip()]
    window = lines[:200]

    # 1) Prefer '##'
    for ln in window:
        if ln.startswith("##"):
            candidate = normalize_markdown_text(ln.lstrip("#"))
            if is_title_like(candidate):
                return candidate
            if candidate:
                return candidate

    # 2) Accept '#'
    for ln in window:
        if ln.startswith("#"):
            candidate = normalize_markdown_text(ln.lstrip("#"))
            if is_title_like(candidate):
                return candidate
            if candidate:
                return candidate

    # 3) Bold-only line: **Title**
    for ln in window:
        if ln.startswith("**") and ln.endswith("**") and len(ln) >= 4:
            candidate = normalize_markdown_text(ln)
            if is_title_like(candidate):
                return candidate

    # 4) First line fallback
    first = window[0] if window else ""
    if is_title_like(first) or (first.isupper() and 8 <= len(first) <= 80):
        return first

    return None
```

```python
records = []

for idx, pdf_path in enumerate(tqdm(pdf_paths, desc="Parsing first pages (md)"), start=1):
    paper_id = pdf_path.stem

    md_text = read_first_page_markdown(pdf_path)
    title = extract_title_from_md(md_text)

    records.append({
        "paper_id": paper_id,
        "title": title if title is not None else "NOT FOUND",
        "first_page_text": md_text,
        "text_len": len(md_text),
    })

    # optional progress log every 10
    if idx % 10 == 0:
        print(f"Parsed {idx}/{len(pdf_paths)} PDFs...")
```

```python
required_cols = {"paper_id", "title", "first_page_text", "text_len"}
assert required_cols.issubset(docs_df.columns), f"docs_df must contain columns: {required_cols}"

# Clean text (avoid None)
docs_df["first_page_text"] = docs_df["first_page_text"].fillna("").astype(str)
docs_df["title"] = docs_df["title"].fillna("").astype(str)

print("✅ docs_df ready:", docs_df.shape)
docs_df.head(3)
```

```python
MODEL_NAME = "BAAI/bge-small-en-v1.5"
embedder = SentenceTransformer(MODEL_NAME)

NORMALIZE_EMB = True

dim = embedder.get_sentence_embedding_dimension()
print("✅ Loaded model:", MODEL_NAME)
print("Embedding dim:", dim)
```

```python
def embed_texts(texts, batch_size=32):
    vectors = []
    for i in tqdm(range(0, len(texts), batch_size), desc="Embedding"):
        batch = texts[i:i+batch_size]
        emb = embedder.encode(
            batch,
            batch_size=batch_size,
            normalize_embeddings=NORMALIZE_EMB,
            show_progress_bar=False
        )
        # ensure float32 for LanceDB
        vectors.append(np.asarray(emb, dtype=np.float32))
    return np.vstack(vectors)

texts = docs_df["first_page_text"].tolist()

embeddings = embed_texts(texts, batch_size=32)
print("✅ embeddings shape:", embeddings.shape)
assert embeddings.shape[0] == len(docs_df) and embeddings.shape[1] == dim
```

```python
DB_DIR = "lancedb_papers"
TABLE_NAME = "papers_first_page"

db = lancedb.connect(DB_DIR)

schema = pa.schema([
    pa.field("paper_id", pa.string()),
    pa.field("title", pa.string()),
    pa.field("first_page_text", pa.string()),
    pa.field("text_len", pa.int32()),
    pa.field("embedding", pa.list_(pa.float32(), dim)),
])

records = []
for i, row in docs_df.iterrows():
    records.append({
        "paper_id": str(row["paper_id"]),
        "title": str(row["title"]),
        "first_page_text": str(row["first_page_text"]),
        "text_len": int(row["text_len"]),
        "embedding": embeddings[i].tolist(),
    })

# Create/overwrite table
if TABLE_NAME in db.table_names():
    db.drop_table(TABLE_NAME)

table = db.create_table(TABLE_NAME, data=records, schema=schema)

print("✅ LanceDB table created.")
print("DB path:", DB_DIR)
print("Table:", TABLE_NAME)
print("Row count:", table.count_rows())
```

```python
def embed_query(q: str) -> np.ndarray:
    v = embedder.encode([q], normalize_embeddings=NORMALIZE_EMB, show_progress_bar=False)
    return np.asarray(v[0], dtype=np.float32)

query = "BERT passage re-ranking"
qvec = embed_query(query)

res = (
    table.search(qvec)
    .limit(5)
    .select(["paper_id", "title", "text_len"])
    .to_pandas()
)

print("Query:", query)
res
```

### بازیابی برداری پرس‌وجو با استفاده از امبدینگ و LanceDB

در این بخش، پرس‌وجوی متنی `"BERT passage re-ranking"` ابتدا توسط مدل امبدینگ به یک بردار عددی در فضای embedding تبدیل می‌شود (`embed_query`). در این فرآیند، تابع `encode` یک بردار چگال (dense vector) تولید می‌کند که در صورت فعال بودن `NORMALIZE_EMB` نرمال‌سازی L2 نیز روی آن انجام می‌شود تا شباهت کسینوسی به‌درستی قابل محاسبه باشد. سپس این بردار به‌عنوان کوئری به پایگاه داده‌ی برداری (`table.search(qvec)`) ارسال شده و نزدیک‌ترین اسناد بر اساس فاصله‌ی برداری بازیابی می‌شوند.

در خروجی نمایش‌داده‌شده:

- ستون `_distance` فاصله‌ی برداری بین پرس‌وجو و هر سند را نشان می‌دهد (مقدار کمتر ⇒ شباهت بیشتر).
- مقاله‌ی `1-1-r` با عنوان *PASSAGE RE-RANKING WITH BERT* کمترین فاصله (≈0.239) را دارد و بنابراین مرتبط‌ترین نتیجه است.
- نتایج بعدی مانند `ColBERT` و `CEDR` نیز به دلیل ارتباط مفهومی با neural ranking بازیابی شده‌اند.
- حضور مقاله‌ی `1-2-n` (مقاله‌ی اصلی BERT) نشان می‌دهد که بازیابی صرفاً بر اساس شباهت معنایی embedding انجام می‌شود و سیستم هنوز فیلتر موضوعی دقیق (مثلاً تمایز بین pretraining و ranking) اعمال نکرده است.

به‌طور کلی، این متد یک **جستجوی برداری مبتنی بر شباهت معنایی (Semantic Vector Search)** را پیاده‌سازی می‌کند که به‌جای تطبیق واژگانی، اسناد را در فضای برداری معنایی رتبه‌بندی می‌کند.

## <div style="text-align: center; direction: rtl; font-family: Vazir;">جستجوی معنایی پایه (۱۰ نمره)</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
در این بخش نیاز است به ازای تمام Initial Queryها در دیتاست کاربران، جستجوی معنایی در پایگاه داده برداری انجام دهید.
سپس مقدار p@1 را برای نتایج جستجو به ازای هر کاربر محاسبه کرده و در انتها این معیار و میانگینش را برای کاربران نمایش دهید.

🎯 <b>خروجی مورد انتظار:</b><br>
جستجوی معنایی برای کوئری‌های اولیه کاربران، محاسبه p@1 و نمایش آن، محاسبه میانگین p@1 برای همه کاربران

```python
USER_ID_COL = "persona_id"
QUERY_COL   = "initial_query"
GT_COL      = "relevant_papers"

for c in [USER_ID_COL, QUERY_COL, GT_COL]:
    assert c in users_df.columns, f"Missing column '{c}'. Available: {list(users_df.columns)}"


NORMALIZE_EMB = True

def embed_query(q: str) -> np.ndarray:
    v = embedder.encode([q], normalize_embeddings=NORMALIZE_EMB, show_progress_bar=False)
    return np.asarray(v[0], dtype=np.float32)

def extract_relevant_paper_ids(relevant_papers) -> set[str]:
    """
    Input can be:
      - list[dict] each containing 'paper_id'
      - list[str]
      - dict
      - string
    Return: set of paper_id strings
    """
    if relevant_papers is None:
        return set()

    if isinstance(relevant_papers, list):
        ids = set()
        for item in relevant_papers:
            if isinstance(item, dict) and "paper_id" in item:
                ids.add(str(item["paper_id"]))
            elif isinstance(item, str):
                ids.add(item.strip())
        return {i for i in ids if i}

    if isinstance(relevant_papers, dict):
        if "paper_id" in relevant_papers:
            return {str(relevant_papers["paper_id"])}
        vals = set()
        for v in relevant_papers.values():
            if isinstance(v, str):
                vals.add(v.strip())
        return {x for x in vals if x}

    if isinstance(relevant_papers, str):
        s = relevant_papers.strip()
        if not s:
            return set()
        if "," in s:
            return {p.strip() for p in s.split(",") if p.strip()}
        return {s}

    return {str(relevant_papers)}

def search_top1(query: str) -> str | None:
    q = (query or "").strip()
    if not q:
        return None
    qvec = embed_query(q)
    res = table.search(qvec).limit(1).select(["paper_id"]).to_pandas()
    if res.empty:
        return None
    return str(res.loc[0, "paper_id"])


rows = []

for _, u in tqdm(users_df.iterrows(), total=len(users_df), desc="Searching + computing P@1"):
    uid = u[USER_ID_COL]
    query = u[QUERY_COL]
    gt_ids = extract_relevant_paper_ids(u[GT_COL])

    top1_id = search_top1(query)
    p_at_1 = int(top1_id is not None and top1_id in gt_ids)

    rows.append({
        "persona_id": uid,
        "initial_query": query,
        "top1_paper_id": top1_id,
        "p@1": p_at_1,
        "num_relevant": len(gt_ids),
    })

p1_df = pd.DataFrame(rows)
mean_p1 = float(p1_df["p@1"].mean()) if len(p1_df) else 0.0

print("✅ Finished.")
print("Users:", len(p1_df))
print("Mean P@1:", mean_p1)

display(p1_df.head(20))
```

```python
USER_SIM_PROMPT = """
You are simulating a researcher-user.

Answer the assistant's question using ONLY the provided information_need JSON.

Rules:
1. Answer ONLY the specific aspect asked in the question.
2. Do NOT mention other constraints unless explicitly asked.
3. Keep the answer short (1 sentence or short bullet list).
4. If the question refers to something not present in information_need, reply exactly:
Not specified in my information need.

Output ONLY the answer text.
""".strip()
```

```python

def simulate_user_answer(
    assistant_question: str,
    information_need: Dict[str, Any],
    *,
    client,
    model: str = "gpt-5",
    temperature: float = 0.2,
) -> str:
    """
    Simulate a user's response to an assistant question using ONLY information_need.
    `client` must already be initialized (as you did with avalai base_url).
    """
    payload = {
        "assistant_question": (assistant_question or "").strip(),
        "information_need": information_need if isinstance(information_need, dict) else {},
    }

    resp = client.chat.completions.create(
        model=model,
        temperature=temperature,
        messages=[
            {"role": "system", "content": USER_SIM_PROMPT},
            {"role": "user", "content": json.dumps(payload, ensure_ascii=False)},
        ],
    )

    return (resp.choices[0].message.content or "").strip()
```

```python
def simulate_user_answer_from_user_row(
    assistant_question: str,
    user_row,   # pandas Series or dict
    *,
    client,
    model: str = "gpt-5",
    temperature: float = 0.2,
) -> str:
    if hasattr(user_row, "to_dict"):
        user_row = user_row.to_dict()

    info_need = user_row.get("information_need", {})
    if isinstance(info_need, str):
        try:
            info_need = json.loads(info_need)
        except Exception:
            info_need = {}

    return simulate_user_answer(
        assistant_question=assistant_question,
        information_need=info_need,
        client=client,
        model=model,
        temperature=temperature,
    )
```

 در این بخش شما باید سیستم جستجوگر مکالمه محور خود را با استفاده از مدل‌های زبانی بزرگ بسازید. به همین منظور باید سیستمی طراحی کنید که به ازای information needهای کاربران از آن‌ها سوالات clarifying بپرسد و با استفاده از تابع شبیه سازی کاربر در مرحله قبل، کاربر پاسخ سوالات را بدهد. سپس باید تاریخچه مکالمه برای هر کاربر را ذخیره کرده و در مراحل بعدی از آن ها استفاده کنید. تعداد دورهای مکالمه و پرسش و پاسخ باید به اندازه ای باشد که کوئری کاربر به اندازه کافی روشن شده باشد و یا همه information needهای خود را به سیستم گفته باشد.

```python
ASSISTANT_SYSTEM_PROMPT = """
You are a conversational academic search assistant.

You are given:
- initial_query
- information_need JSON (this contains ALL valid constraints)

Rules:
1. Ask ONLY about fields that exist in information_need.
2. Ask about ONE field at a time.
3. Do NOT invent new constraints or sub-constraints.
4. Once all fields in information_need are clarified, stop asking deeper questions.
5. Ask a short, focused question (one sentence).

Valid fields:
- specific_topic
- sub_topics
- year_range
- paper_type
- preferred_venues
- specific_aspects
- exclude

Output ONLY the clarification question.
""".strip()
```

```python
SLOT_ORDER = [
    "specific_topic",
    "sub_topics",
    "year_range",
    "paper_type",
    "preferred_venues",
    "specific_aspects",
    "exclude",
]

SLOT_QUESTION_HINTS = {
    "specific_topic": "Ask what specific topic within the broad area they mean.",
    "sub_topics": "Ask which sub-topics/approaches they care about.",
    "year_range": "Ask which years/time range they want.",
    "paper_type": "Ask what kind of papers (conference/journal/survey/etc.).",
    "preferred_venues": "Ask which venues/conferences/journals they prefer.",
    "specific_aspects": "Ask which aspects to emphasize (efficiency, robustness, domain adaptation, etc.).",
    "exclude": "Ask what to exclude / what they do not want.",
}

def normalize_info_need(x: Any) -> Dict[str, Any]:
    if isinstance(x, dict):
        return x
    if isinstance(x, str):
        try:
            y = json.loads(x)
            return y if isinstance(y, dict) else {}
        except Exception:
            return {}
    return {}

def slot_is_filled(info_need: Dict[str, Any], slot: str) -> bool:
    if slot not in info_need:
        return False
    v = info_need.get(slot)
    if v is None:
        return False
    if isinstance(v, str) and not v.strip():
        return False
    if isinstance(v, list) and len(v) == 0:
        return False
    return True

def get_remaining_slots(info_need: Dict[str, Any], asked_slots: set[str]) -> List[str]:
    remaining = []
    for s in SLOT_ORDER:
        if slot_is_filled(info_need, s) and s not in asked_slots:
            remaining.append(s)
    return remaining
```

```python
def llm_generate_clarifying_question(
    *,
    initial_query: str,
    information_need: Dict[str, Any],
    conversation: List[Dict[str, str]],
    remaining_slots: List[str],
    client,
    model: str = "gpt-5",
    temperature: float = 0.2,
) -> str:
    slot_hints = [SLOT_QUESTION_HINTS[s] for s in remaining_slots] if remaining_slots else []

    payload = {
        "initial_query": initial_query,
        "information_need": information_need,
        "remaining_slots": remaining_slots,
        "slot_hints": slot_hints,
        "conversation_so_far": conversation[-6:],
    }

    resp = client.chat.completions.create(
        model=model,
        temperature=temperature,
        messages=[
            {"role": "system", "content": ASSISTANT_SYSTEM_PROMPT},
            {"role": "user", "content": json.dumps(payload, ensure_ascii=False)},
        ],
    )
    q = (resp.choices[0].message.content or "").strip()
    q = q.split("\n")[0].strip()
    if not q.endswith("?"):
        q += "?"
    return q
```

```python
def infer_asked_slot_from_question(question: str, remaining_slots: List[str]) -> str | None:
    q = question.lower()
    rules = [
        ("year_range", ["year", "years", "time", "between", "from", "to", "recent", "older"]),
        ("preferred_venues", ["venue", "venues", "conference", "journal", "sigir", "acl", "emnlp", "neurips", "icml"]),
        ("paper_type", ["paper type", "workshop", "survey", "tutorial", "journal", "conference"]),
        ("specific_aspects", ["aspect", "focus", "emphasize", "efficiency", "domain", "adaptation", "latency"]),
        ("exclude", ["exclude", "avoid", "not interested", "do not want"]),
        ("sub_topics", ["subtopic", "approach", "methods", "models", "cross-encoder", "bert", "rerank", "ranker"]),
        ("specific_topic", ["specific", "exactly", "which topic", "what do you mean", "narrow down"]),
    ]
    for slot, kws in rules:
        if slot in remaining_slots and any(k in q for k in kws):
            return slot
    if len(remaining_slots) == 1:
        return remaining_slots[0]
    return None
```

```python
def run_conversation_for_user(
    user_row,
    *,
    client,
    assistant_model: str = "gpt-5",
    user_model: str = "gpt-5",
    max_turns: int = 4,
) -> Dict[str, Any]:

    user = user_row.to_dict() if hasattr(user_row, "to_dict") else dict(user_row)
    persona_id = user.get("persona_id")
    initial_query = str(user.get("initial_query", "")).strip()
    info_need = normalize_info_need(user.get("information_need", {}))

    conversation: List[Dict[str, str]] = []
    asked_slots: set[str] = set()

    # seed conversation with the user's initial query
    conversation.append({"role": "user", "content": initial_query})

    for _ in range(max_turns):
        remaining = get_remaining_slots(info_need, asked_slots)
        if not remaining:
            break

        assistant_q = llm_generate_clarifying_question(
            initial_query=initial_query,
            information_need=info_need,
            conversation=conversation,
            remaining_slots=remaining,
            client=client,
            model=assistant_model,
        )
        conversation.append({"role": "assistant", "content": assistant_q})

        user_a = simulate_user_answer_from_user_row(
            assistant_q,
            user_row,
            client=client,
            model=user_model,
        )
        conversation.append({"role": "user", "content": user_a})

        slot = infer_asked_slot_from_question(assistant_q, remaining)
        if slot:
            asked_slots.add(slot)

    conversation_text = "\n".join([f"{m['role'].upper()}: {m['content']}" for m in conversation])

    return {
        "persona_id": persona_id,
        "initial_query": initial_query,
        "turns": sum(1 for m in conversation if m["role"] == "assistant"),
        "conversation_messages": conversation,
        "conversation_text": conversation_text,
    }
```

در این بخش یک **سیستم مکالمه‌ی مرحله‌ای (Conversational Clarification Loop)** طراحی شده است که نقش یک دستیار جستجوی آکادمیک را ایفا می‌کند. هدف این سیستم، تبدیل یک کوئری کلی (مانند *"neural networks"*) به یک نیاز اطلاعاتی دقیق و ساخت‌یافته از طریق پرسش‌های شفاف‌ساز است.

منطق کلی راهکار به این صورت است:
- دستیار فقط درباره‌ی فیلدهایی سؤال می‌پرسد که در `information_need` وجود دارند.
- در هر مرحله فقط درباره‌ی یک اسلات (slot) سؤال می‌شود.
- پس از هر سؤال، پاسخ کاربر با استفاده از **User Simulator** (تعریف‌شده در بخش قبل) تولید می‌شود.
- این فرآیند تا پر شدن اسلات‌ها یا رسیدن به `max_turns` ادامه می‌یابد.

در تست اجراشده برای Persona 1:
- کوئری اولیه کلی است: `neural networks`
- دستیار به‌صورت مرحله‌ای موارد زیر را شفاف می‌کند:
  1. موضوع دقیق‌تر (Neural ranking models for passage retrieval)
  2. زیرموضوع‌ها (BERT-based rankers, Cross-encoders)
  3. نوع مقاله (Conference papers)
  4. موارد قابل حذف (Exclude pure theoretical papers)

در نهایت یک مکالمه‌ی ۴ مرحله‌ای تولید شده که تمام قیود موجود در `information_need` را پوشش می‌دهد.

سپس همین فرآیند برای تمام کاربران اجرا شده و:
- یک دیتافریم شامل مکالمات ساخته شده

نتیجه جستجوی معنایی با استفاده از مکالمه کامل هر کاربر و گزارش معیار p@1 برای هر کاربر و میانگین این معیار برای تمامی کاربران.

```python
LANCEDB_DIR = "lancedb_papers"
TABLE_NAME = "papers_first_page"

embedder = SentenceTransformer("BAAI/bge-small-en-v1.5")

db = lancedb.connect(LANCEDB_DIR)
tbl = db.open_table(TABLE_NAME)

print("✅ LanceDB connected:", LANCEDB_DIR)
print("✅ Table opened:", TABLE_NAME)
```

```python
users_map = {int(r["persona_id"]): r for _, r in users_df.iterrows()}

results = []

for _, conv_row in tqdm(conversations_df.iterrows(),
                         total=len(conversations_df),
                         desc="Conversation Semantic Search"):

    persona_id = int(conv_row["persona_id"])
    conv_text = str(conv_row["conversation_text"]).strip()

    user_row = users_map.get(persona_id)
    if user_row is None:
        continue

    relevant_ids = extract_relevant_paper_ids(
        user_row.get("relevant_papers")
    )

    query_vec = embedder.encode(
        conv_text,
        normalize_embeddings=True
    ).astype(np.float32)

    top_df = (
        tbl.search(query_vec)
            .limit(1)
            .to_pandas()
    )

    if top_df.empty:
        top1_id = None
        p1 = 0
        score = None
    else:
        top1_id = str(top_df.iloc[0]["paper_id"])
        score = top_df.iloc[0].get("_distance", None)
        p1 = 1 if top1_id in relevant_ids else 0

    results.append({
        "persona_id": persona_id,
        "top1_paper_id": top1_id,
        "score": score,
        "p_at_1": p1,
        "num_relevant": len(relevant_ids)
    })

results_df = pd.DataFrame(results)

mean_p1 = results_df["p_at_1"].mean()

print("✅ Done.")
print("Mean P@1:", mean_p1)

display(results_df.head())
```

<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی‌های مورد انتظار:</b><br><br>
تعریف صحیح ایجنت و سیستم جستجوی مکالمه‌ای و چاپ و ذخیره کوئری‌های بازنویسی شده.

```python

def format_conversation(conversation: list[dict]) -> str:
    """
    conversation: list of {"role": "user"/"assistant", "content": "..."}
    returns a compact transcript suitable for LLM input.
    """
    lines = []
    for m in conversation:
        role = m.get("role", "").upper()
        content = (m.get("content") or "").strip()
        if not content:
            continue
        lines.append(f"{role}: {content}")
    return "\n".join(lines).strip()

def _clean_one_line(s: str) -> str:
    s = (s or "").strip()
    s = re.sub(r"\s+", " ", s)
    return s

def llm_text(messages, temperature=0.2) -> str:
    resp = client.chat.completions.create(
        model="gpt-5",
        messages=messages,
        temperature=temperature,
    )
    out = (resp.choices[0].message.content or "").strip()
    return out
```

```python
REWRITE_SYSTEM_PROMPT = """
You are a query rewriting agent for academic paper search.
Rewrite the user's initial query using the conversation transcript.

Constraints:
- Output MUST be a single line.
- Output MUST be concise (max ~20 words).
- Include key constraints only if clearly stated in the conversation (topic, subtopic, year range, venues, exclusions).
- Do NOT add any new constraints that are not in the transcript.
- Do NOT include quotes, bullets, or explanations. Only the rewritten query line.
"""

def rewrite_query_from_conversation(initial_query: str, conversation: list[dict]) -> str:
    transcript = format_conversation(conversation)

    messages = [
        {"role": "system", "content": REWRITE_SYSTEM_PROMPT.strip()},
        {"role": "user", "content": f"Initial query: {initial_query}\n\nConversation:\n{transcript}\n\nRewritten query:"}
    ]

    out = llm_text(messages, temperature=0.2)
    out = _clean_one_line(out)

    if not out:
        return _clean_one_line(initial_query)

    out = out.splitlines()[0].strip()
    out = _clean_one_line(out)

    if len(out) > 220:
        out = out[:220].rstrip()

    return out
```

```python
merged_df = users_df.merge(
    conversations_df[["persona_id", "conversation_messages"]],
    on="persona_id",
    how="left"
)

rows = []

for _, u in tqdm(merged_df.iterrows(),
                 total=len(merged_df),
                 desc="Rewriting queries"):

    pid = int(u["persona_id"])
    initial_query = u["initial_query"]
    conv = u["conversation_messages"] if isinstance(u["conversation_messages"], list) else []

    rewritten = rewrite_query_from_conversation(initial_query, conv)

    rows.append({
        "persona_id": pid,
        "initial_query": initial_query,
        "rewritten_query": rewritten,
        "turns": sum(1 for m in conv if (m.get("content") or "").strip()),
        "conversation_text": format_conversation(conv),
    })

rewritten_df = pd.DataFrame(rows).sort_values("persona_id").reset_index(drop=True)

print("✅ Rewritten queries created.")
display(rewritten_df[["persona_id", "initial_query", "rewritten_query", "turns"]].head())
```

## <div style="text-align: center; direction: rtl; font-family: Vazir;">جستجوی معنایی با کوئری‌های بازنویسی شده (۵ نمره)</div>


<div dir="rtl" style="
  text-align: right;
  padding: 20px;
  background-color: #6b7280;
  border-radius: 12px;
  border: 2px solid #022216;
  font-family: 'Vazir','Tahoma',sans-serif;
  margin-bottom: 30px;
">
  <p style="line-height: 1.8; margin:0;">
در این بخش باید با استفاده از پایگاه داده برداری که در مراحل قبل ایجاد کرده‌اید جستجوی معنایی را بر اساس کوئری‌های بازنویسی شده از مکالمات کاربران انجام دهید. سپس با توجه به نتایج و برچسب مقالات مرتبط هر کاربر در دیتاست کاربران،‌ معیار p@1 برای هر کاربر و میانگین این معیار را برای همه کاربران محاسبه و گزارش کنید.
  </p>




<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی‌های مورد انتظار:</b><br><br>
نتیجه جستجوی معنایی با استفاده از کوئری بازنویسی شده هر کاربر و گزارش معیار p@1 برای هر کاربر و میانگین این معیار برای تمامی کاربران.
</ul>
</p>


```python
results = []

for _, row in tqdm(rewritten_df.iterrows(),
                   total=len(rewritten_df),
                   desc="Semantic Search (Rewritten Queries)"):

    pid = int(row["persona_id"])
    rewritten_query = str(row["rewritten_query"]).strip()

    user_row = users_map.get(pid)
    if user_row is None:
        continue

    relevant_ids = extract_relevant_paper_ids(
        user_row.get("relevant_papers")
    )

    q_vec = embedder.encode(
        rewritten_query,
        normalize_embeddings=True
    ).astype(np.float32)

    top_df = (
        tbl.search(q_vec)
            .limit(1)
            .to_pandas()
    )

    if top_df.empty:
        top1_id = None
        score = None
        p1 = 0
    else:
        top1_id = str(top_df.iloc[0]["paper_id"])
        score = top_df.iloc[0].get("_distance", None)
        p1 = 1 if top1_id in relevant_ids else 0

    results.append({
        "persona_id": pid,
        "rewritten_query": rewritten_query,
        "top1_paper_id": top1_id,
        "score": score,
        "num_relevant": len(relevant_ids),
        "p_at_1": p1
    })

rewritten_search_df = pd.DataFrame(results).sort_values("persona_id").reset_index(drop=True)

mean_p1 = rewritten_search_df["p_at_1"].mean()

print("✅ Done.")
print("📊 Mean P@1 (rewritten queries):", mean_p1)

display(rewritten_search_df.head(10))
```
